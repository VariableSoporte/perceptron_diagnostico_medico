## **Normalización**
Existen varias técnicas de normalización sin embargo una de las mas comunes para aplicar en perceptrones es la **Normalización Min-Max**. Esta técnica transforma cualquier valor para que quede en el rango de 0 a 1.

Su formula es:


$$x_{norm} = \frac{x - x_{min}}{x_{max} - x_{min}}$$
